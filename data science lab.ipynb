```python
import numpy as np

# 1. Custom User-Defined function for Euclidean Distance
def calculate_euclidean_distance(point1, point2):
    
    p1 = np.array(point1)
    p2 = np.array(point2)
    
    # Formula apply 
    distance = np.sqrt(np.sum((p1 - p2) ** 2))
    return distance

# 2. Test points (X aur Y values)
row1 = [44.97, -103.77]
row2 = [62.40, 96.80]

# 3. Distance calculate 
result = calculate_euclidean_distance(row1, row2)
print(f"Euclidean Distance between points is: {result:.2f}")
```

```python
import numpy as np

# Data points
data = np.array([
    [10, 2],   # P1
    [12, 4],   # P2
    [16, 6],   # P3
    [20, 8],   # P4
    [24, 10],   # P5
    [26, 12],   # P6
    [30, 14],   # P7
    [34, 16]    # P8
])

# User-defined Euclidean Distance function
def euclidean_distance(p1, p2):
    return np.sqrt((p2[0] - p1[0])**2 + (p2[1] - p1[1])**2)

# Consecutive points distance
print("--- Consecutive Points Distance ---")

for i in range(len(data) - 1):
    dist = euclidean_distance(data[i], data[i+1])
    print(f"Distance between P{i+1} and P{i+2}: {dist:.2f}")
```

```python
import numpy as np
import math

# Data points
data = np.array([
    [10, 2],    # P1
    [12, 4],    # P2
    [16, 6],    # P3
    [20, 8],    # P4
    [24, 10],   # P5
    [26, 12],   # P6
    [30, 14],   # P7
    [34, 16]    # P8
])

# K-Means clustering
k = 2

# Initial centroids
c1 = np.array([10, 2])
c2 = np.array([34, 16])

for i in range(5):

    cluster1 = []
    cluster2 = []

    # Calculating distance
    for point in data:

        d1 = math.sqrt(
            (point[0] - c1[0])**2 +
            (point[1] - c1[1])**2
        )

        d2 = math.sqrt(
            (point[0] - c2[0])**2 +
            (point[1] - c2[1])**2
        )

        if d1 < d2:
            cluster1.append(point)
        else:
            cluster2.append(point)

    # Calculating new centroid
    c1 = np.mean(cluster1, axis=0)
    c2 = np.mean(cluster2, axis=0)

print("Cluster 1:", cluster1)
print("Cluster 2:", cluster2)

print("Centroid 1:", c1)
print("Centroid 2:", c2)
```

```python
# plotting clusters

plt.scatter(np.array(cluster1)[:,0], np.array(cluster1)[:,1])
plt.scatter(np.array(cluster2)[:,0], np.array(cluster2)[:,1])

plt.scatter(c1[0], c1[1], marker='x', s=100)
plt.scatter(c2[0], c2[1], marker='x', s=100)

plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```
