```python

import matplotlib.pyplot as plt
from scipy import stats

# Data points
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

# Linear regression fit 
slope, intercept, r, p, std_err = stats.linregress(x, y)


def myfunc(x_val):
    return slope * x_val + intercept


# User se input 
user_input = float(input("Linear Regression ke liye x ki value dalein: "))
predicted_y = myfunc(user_input)
print(f"Predicted y value: {predicted_y}")
```

# polynomial

```python
import numpy as np

# Data points
x = [1, 2, 3, 5, 6, 7, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [100, 90, 80, 60, 60, 55, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Polynomial model (degree 3)
mymodel = np.poly1d(np.polyfit(x, y, 3))

# User se input lena aur predict karna
user_input = float(input("Polynomial Regression ke liye x ki value dalein: "))
predicted_y = mymodel(user_input)
print(f"Predicted y value: {predicted_y}")
```
