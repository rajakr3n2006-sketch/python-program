```python
import numpy as np

a = np.arange(1, 13)

print(a)

b = a.reshape(3, 4)

print(b)
print("Maximum:", b.max())
print("Minimum:", b.min())
print("Sum:", b.sum())
print("Mean:", b.mean())
print("Shape:", b.shape)
print("Size:", b.size)
```

```python
import numpy as np

asearch = np.array([2, 3, 6, 9, 10])

abc = int(input("Enter the element to search: "))
print(abc)

res = np.where(asearch == abc)

print("Element found in index:", res)
```
